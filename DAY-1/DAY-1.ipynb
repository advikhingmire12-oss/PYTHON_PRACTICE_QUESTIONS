# There is a single operator in Python, capable of providing the remainder of a division operation. Two numbers are passed as parameters. The first parameter divided by the second parameter will have a remainder, possibly zero. Return that value.

```python
def remainder(x, y):
    division = x % y
    if division == 0:
        print("x is divisible by y and returns 0 as remainder")
    else:
        print("x is not divisible by y")
    return division
print(remainder(7,2))
```

# The starter below has a bug. Fix lowest_reading. Return the smallest reading in readings.

```python
def lowest_reading(readings):
    lowest = readings[0]
    for reading in readings:
        if reading < lowest:    
            lowest = reading
    return lowest
lowest_reading([5,4,232])
```

# You are checking your Apple devices before heading out. For this challenge, a device needs charging when its battery is below 20%. Create needs_charging to return the names of those devices, in their original order. Each entry in devices is a pair: [name, battery].

```python
def needs_charging(devices):
    low_battery = []
    for name, battery in devices:
        if battery < 20:
            low_battery.append(name)
    return low_battery
needs_charging([["iphone",18],["mackbook",18]])
```
